## 2. The question: decision, action, cost of a wrong call

**Decision:** which pages a reviewer should look at first this week, when there are way more pages than they have time to check.

**Who acts, and what they do:** a content reviewer opens the ranked list and, for each top page, either refreshes it, fixes the title/metadata, leaves it alone because it's fine, or deprioritizes it if there's not enough demand to justify the time.

**Cost of a wrong call:** two different mistakes, and they cost different things. Flagging a page that's actually fine wastes a reviewer's limited hours — time that could've gone somewhere it mattered. Missing a page that's genuinely declining is worse, because the traffic loss keeps quietly compounding until someone notices. Since a reviewer can only realistically get through maybe 20-50 pages a week out of a much bigger inventory, getting the *top* of the list right matters more than being right about every page overall.

## 3. Quick look at the data (2-3 real numbers)

Quick note before the code: loading the starter CSV to check whether there are actually enough real candidates here to justify spending 7 weeks on this lane, and whether the existing rule-based flags are catching different things or just the same pages over and over.


```python
# Clone the repo (skip this if you've already cloned it in this session)
!git clone -q https://github.com/zainabkhalid663/Flyrank-ML-Internship.git

import pandas as pd

df = pd.read_csv("Flyrank-ML-Internship/data/raw/content_refresh_anonymized.csv")

total = len(df)
n_clients = df["client_id"].nunique()

declining_with_demand = ((df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)).sum()
page_one_decay_risk = ((df["avg_position"] > 0) & (df["avg_position"] <= 10) & (df["content_age_days"] >= 180)).sum()
low_ctr_visible_page = ((df["impressions_90d"] >= 500) & (df["avg_position"] > 0) & (df["avg_position"] <= 20) & (df["ctr"] < 0.5)).sum()

print(f"rows: {total:,} | clients: {n_clients}")
print(f"declining_with_demand candidates: {declining_with_demand:,} ({declining_with_demand/total:.1%} of inventory)")
print(f"page_one_decay_risk candidates:   {page_one_decay_risk:,} ({page_one_decay_risk/total:.1%} of inventory)")
print(f"low_ctr_visible_page candidates:  {low_ctr_visible_page:,} ({low_ctr_visible_page/total:.1%} of inventory)")
```

## 4. Careful words: what I can and can't claim

**What I can claim:**
- *Observed* patterns — pages with a certain mix of signals showed up more often in the declining or low-CTR groups, based on what actually happened in this sample.
- *Directional* signals — things like "low CTR at a strong position tends to mean under-captured clicks," without claiming to know the exact cause.
- *Decision-support* results — a ranked list that helps a reviewer spend limited time on the most promising pages first, tested with precision@K, not a promise about any single page.

**What I won't claim:**
- That a refresh **causes** a page to recover — I have no experiment set up for that, so any before/after comparison is a hint at best, never proof.
- That I've found a **Google ranking factor** — I only have FlyRank's own metrics (impressions, clicks, position, CTR), not Google's actual algorithm.
- That the starter's label (`trend_direction == "down"`) is a real future prediction — it's just a snapshot of the current window, and I'll say so clearly if I end up keeping it instead of building a proper prior-90-days → next-30-days label.
- That "worth reviewing" means "guaranteed to improve" — the model just points a human toward where to look, it doesn't replace their judgment.
